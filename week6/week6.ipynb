```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
'''1. Load the IPL dataset and display first 10 records.
Find the total number of:
• players
• Teams
• Seasons
Plot the number of players played per season using Matplotlib.
Use Seaborn to visualize:
• Top 5 players with most runs
Create a bar plot showing top 10 wicket takers.
Plot a pie chart representing for players per country
. Compare any 2 parameters using an appropriate plot.
Create a visualization that tells a clear IPL performance story (your choice).
Add annotations, legends, and titles to improve readability'''
```

```python
ipl = pd.read_csv("ipl.csv")
ipl.head(10)
```

```python
total_players = ipl.shape[0]
total_teams = ipl['TEAM'].nunique()
total_seasons = ipl['AUCTION YEAR'].nunique()

print("Players:", total_players)
print("Teams:", total_teams)
print("Seasons:", total_seasons)
```

```python
players_per_season = ipl['AUCTION YEAR'].value_counts().sort_index()

plt.figure(figsize=(10,6))
players_per_season.plot(kind='bar', color='skyblue')
plt.title("Number of Players per Season")
plt.xlabel("Season (Auction Year)")
plt.ylabel("Number of Players")
plt.show()
```

```python
top5_runs = ipl[['PLAYER NAME','T-RUNS']].sort_values(by='T-RUNS', ascending=False).head(5)

plt.figure(figsize=(8,6))
sns.barplot(x=top5_runs['PLAYER NAME'], y=top5_runs['T-RUNS'], palette="viridis")
plt.title("Top 5 Players with Most Runs")
plt.ylabel("Runs")
plt.xlabel("Players")
plt.xticks(rotation=45)
plt.show()
```

```python
top10_wickets = ipl[['PLAYER NAME','T-WKTS']].sort_values(by='T-WKTS', ascending=False).head(10)

plt.figure(figsize=(10,6))
sns.barplot(x=top10_wickets['PLAYER NAME'], y=top10_wickets['T-WKTS'], palette="cividis")
plt.title("Top 10 Wicket Takers", fontsize=16)
plt.ylabel("Wickets", fontsize=12)
plt.xlabel("Players", fontsize=12)
plt.xticks(rotation=45)

plt.show()
```

```python
country_counts = ipl['COUNTRY'].value_counts()

plt.figure(figsize=(8,8))
plt.pie(country_counts.values, labels=country_counts.index, autopct='%1.1f%%')
plt.title("Players by Country")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x=ipl['T-RUNS'], y=ipl['SIXERS'], hue=ipl['TEAM'], alpha=0.7)
plt.title("Comparison of Runs vs Sixers")
plt.xlabel("Total Runs")
plt.ylabel("Sixes")
plt.legend(title="Team")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x=ipl['BASE PRICE'], y=ipl['SOLD PRICE'], hue=ipl['AUCTION YEAR'], palette="tab10")
plt.title("Auction Trends: Base Price vs Sold Price")
plt.xlabel("Base Price")
plt.ylabel("Sold Price")
plt.legend(title="Auction Year")
plt.show()
```
