```python
import seaborn as sns
data1 = pd.read_csv("mtcars.csv")

print("\nDataset Columns:")
print(data1.columns)

print("\nDataset Rows:")
print(data1[:])

print("\nDataset Types:")
print(data1.dtypes)


mpg = data1['mpg']
hp = data1['hp']
wt = data1['wt']


print("\nStatistics for Miles Per Gallon")
print("Mean: ",mpg.mean())
print("Median: ",mpg.median())
print("Mode: ",mpg.mode())
print("Min: ",mpg.min())
print("Max: ",mpg.max())
print()
print("Statistics for hp")
print("Mean: ",hp.mean())
print("Median: ",hp.median())
print("Mode: ",hp.mode())
print("Min: ",hp.min())
print("Max: ",hp.max())
print()
print("Statistics for wt")
print("Mean: ",wt.mean())
print("Median: ",wt.median())
print("Mode: ",wt.mode())
print("Min: ",wt.min())
print("Max: ",wt.max())


sns.boxplot(x='cyl', y='mpg', data=data1)
plt.title("Fuel Efficiency (mpg) by Number of Cylinders")
plt.xlabel("Cylinders")
plt.ylabel("Miles per Gallon")
plt.show()
```

```python
plt.scatter(data1['hp'],data1['mpg'])
plt.title("Horsepower vs Fuel Efficiency")
plt.xlabel("Horsepower")
plt.ylabel("Miles per Gallon")
plt.show()
```

```python
sns.boxplot(x='am', y='mpg', data=data1)
plt.title("MPG Comparison: Automatic vs Manual Transmission")
plt.xlabel("Transmission (0=Auto, 1=Manual)")
plt.ylabel("Miles per Gallon")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x='hp', y='mpg', data=data1)

for i in range(data1.shape[0]):
    plt.text(data1.hp[i]+2, data1.mpg[i], data1.index[i], fontsize=8)

plt.title("Car Models: Balancing Power and Fuel Efficiency")
plt.xlabel("Horsepower")
plt.ylabel("Miles per Gallon")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

cereals = pd.read_csv("Cereals.csv")
cereals['sugars'] = cereals['sugars'].replace('?', pd.NA)
cereals['sugars'] = pd.to_numeric(cereals['sugars'])
cereals = cereals.dropna(subset=['sugars'])

print(cereals.shape)
print(cereals.dtypes)
print(cereals.isnull().sum())
```

```python
print("Average Calories:", cereals['calories'].mean())
print("Average Sugar:", cereals['sugars'].mean())
print("Average Fiber:", cereals['fiber'].mean())
```

```python
top_sugar = cereals.nlargest(5, 'sugars')[['name','sugars']]
plt.bar(top_sugar['sugars'], top_sugar['name'])
plt.title("Top 5 Cereals by Sugar Content")
plt.xlabel("Sugar (g)")
plt.ylabel("Cereal")
plt.show()
```

```python
plt.scatter(cereals['rating'], cereals['calories'])
plt.xlabel("Rating")
plt.ylabel("Calories")
plt.title("Rating v Calories Comparison")
plt.show()
```

```python
plt.scatter(cereals['calories'], cereals['rating'])
plt.title("Calories vs Consumer Rating")
plt.xlabel("Calories")
plt.ylabel("Rating")
plt.show()
```

```python
sns.boxplot(x='type', y='calories', data=cereals)
plt.title("Calorie Distribution: Cold vs Hot Cereals")
plt.xlabel("Type (C=Cold, H=Hot)")
plt.ylabel("Calories")
plt.show()
```

```python
healthy = cereals[(cereals['sugars'] <= 5) & (cereals['fiber'] >= 3)]
print("Healthier cereals based on sugar ≤ 5g and fiber ≥ 3g:")
print(healthy[['name','sugars','fiber','calories','rating']])
plt.scatter(cereals['sugars'],cereals['fiber'])
plt.title("Sugar vs Fiber: Identifying Healthier Cereals")
plt.xlabel("Sugar (g)")
plt.ylabel("Fiber (g)")
plt.show()
```
